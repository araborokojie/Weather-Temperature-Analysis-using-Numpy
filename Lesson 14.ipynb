### Part 1- Create the Dataset

```python
temperatures = [27.5, 29.0, 30.5, 28.0, 31.0, 32.5, 30.0]

temperature_array = np.array(temperatures)

temperature_array
```

### Part 2- Understand Your Array

```python
print("Number of dimensions:", temperature_array.ndim)
print("Shape:", temperature_array.shape)
print("Data type:", temperature_array.dtype)
print("Number of observations:", temperature_array.size)
```

```python
# Therefore, it is a 1D array, and a 1D NumPy array is commonly described as a vector.
# This also means that it is a one dimension containing 7 values. 
# The data type is float64 and there are 7 temperature observations.
```

### Part 3- Perform Numerical Operations

```python
# create the target array
target = [28.0, 28.0, 29.0, 28.0, 30.0, 30.0, 29.0]

target_array = np.array(target)

print("Target Array:",target_array)

# Temperature Difference
temperature_difference = temperature_array - target_array

print("Temperature Difference:",temperature_difference)

# Increase Every Temperature by 2°C
increased_temperatures = temperature_array + 2

increased_temperatures
print("Increased Temperature:",increased_temperatures)
```

### Part 4- Think Like an Analyst

```python
 # Which day appears to have the highest recorded temperature?
temperature_array.max()
temperature_array.argmax()
print("Highest Temperature:",temperature_array.max())
print("Position of Highest Temperature:",temperature_array.argmax())

# On which days was the recorded temperature above the target?
# True means the recorded temperature was above the target.
temperature_array > target_array
print("Days above the Target:",temperature_array > target_array)

# What does the temperature difference tell the weather-monitoring team?.
```

### What does the temperature difference tell the weather-monitoring team?

The temperature difference shows how much the recorded temperature was above or below the target temperature for each day. A positive value means the recorded temperature was above the target, a negative value means it was below the target, and zero means it matched the target.


### Why would using a NumPy array be more convenient than keeping the temperatures only as a Python list?
NumPy arrays make numerical analysis more convenient because mathematical operations can be performed directly on all the values element by element. This makes it easy to calculate differences, increase values, compare data, and perform other numerical operations without manually processing each value with a loop.

# MINI PROJECT: Weather Temperature Analysis Using NumPy

### Project Introduction

This project analyzes temperature data collected from three weather-monitoring
locations: Lagos, Abuja, and Ibadan. NumPy is used to convert the raw Python
lists into numerical arrays and perform calculations and comparisons.
The analysis will identify the location with the highest temperature on each
day, calculate temperatures after a 1°C increase, and compare temperature
readings between Lagos and Abuja.

### Dataset

The weather station recorded temperatures for five days at three locations:
Lagos, Abuja, and Ibadan.

All temperatures are measured in degrees Celsius (°C).

```python
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]

abuja = [25.0, 28.0, 29.5, 30.0, 32.0]

ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

lagos_array = np.array(lagos)

abuja_array = np.array(abuja)

ibadan_array = np.array(ibadan)


print("Lagos:", lagos_array)
print("Abuja:", abuja_array)
print("Ibadan:", ibadan_array)
```

```python
# Array Inspection
print("Lagos dimensions:", lagos_array.ndim)
print("Lagos shape:", lagos_array.shape)
print("Lagos data type:", lagos_array.dtype)
print("Number of Lagos observations:", lagos_array.size)
print()
print("Abuja dimensions:", abuja_array.ndim)
print("Abuja shape:", abuja_array.shape)
print("Abuja data type:", abuja_array.dtype)
print("Number of Abuja observations:", abuja_array.size)
print()
print("Ibadan dimensions:", ibadan_array.ndim)
print("Ibadan shape:", ibadan_array.shape)
print("Ibadan data type:", ibadan_array.dtype)
print("Number of Ibadan observations:", ibadan_array.size)
```

```python
# Challenge 1 — Find the Highest Temperature on Each Day
all_temperatures = np.array([
    lagos_array,
    abuja_array,
    ibadan_array
])

print ("All Temperatures:",all_temperatures)

highest_each_day = np.max(all_temperatures, axis=0)
print("Highest Temperature for each day:",highest_each_day)


highest_location_index = np.argmax(all_temperatures, axis=0)
print("Highest Location:",highest_location_index)

# axis=0 tells NumPy to compare the values down each column
```

```python
# Challenge 2 — Increase Every Temperature by 1°C
lagos_plus_1 = lagos_array + 1
abuja_plus_1 = abuja_array + 1
ibadan_plus_1 = ibadan_array + 1

print("Lagos after +1°C:", lagos_plus_1)
print("Abuja after +1°C:", abuja_plus_1)
print("Ibadan after +1°C:", ibadan_plus_1)
```

```python
#Challenge 3 — Compare Lagos With Abuja
lagos_abuja_difference = lagos_array - abuja_array

print("Lagos-Abuja Difference:",lagos_abuja_difference)

lagos_array > abuja_array

#Lagos vs Abuja Comparison :The temperature difference was calculated by subtracting Abuja's temperature from Lagos' temperature for each day.
#A positive difference means Lagos was warmer than Abuja, while a negative difference means Lagos was cooler than Abuja.
```

### Findings

1. The temperature data shows that Lagos recorded the highest temperature on
   Days 1, 2, and 3, while Abuja recorded the highest temperature on Days 4
   and 5.

2. Abuja recorded the highest temperature in the dataset, reaching 32.0°C on
   Day 5.

3. Lagos was warmer than Abuja on Days 1, 2, and 3, but Abuja was warmer than
   Lagos on Days 4 and 5.

4. The largest temperature difference between Lagos and Abuja occurred on
   Day 1, when Lagos was 2.5°C warmer than Abuja.

5. Adding 1°C to the arrays demonstrates that NumPy can perform the same
   numerical operation on every element without requiring a for loop.
